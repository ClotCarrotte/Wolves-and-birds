```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Charger les données
tableau = pd.read_csv("occurrence.txt", sep="\t", low_memory=False)

# Isoler les 10 espèces les plus fréquentes
top_10 = tableau['scientificName'].value_counts().head(10)

# Créer le graphique
plt.figure(figsize=(10, 6))
sns.barplot(y=top_10.index, x=top_10.values, palette="viridis")
plt.title("Top 10 des espèces les plus observées")
plt.xlabel("Nombre d'observations")
plt.ylabel("Espèce")
plt.tight_layout()
plt.show()
```

```python
# S'assurer que la colonne 'month' est bien numérique
tableau['month'] = pd.to_numeric(tableau['month'], errors='coerce')

plt.figure(figsize=(10, 6))
sns.countplot(data=tableau, x='month', palette="magma")
plt.title("Répartition des observations par mois")
plt.xlabel("Mois de l'année")
plt.ylabel("Nombre d'observations")
plt.tight_layout()
plt.show()
```

```python
# Filtrer les lignes sans coordonnées GPS
obs_gps = tableau.dropna(subset=['decimalLatitude', 'decimalLongitude'])

plt.figure(figsize=(8, 8))
# L'argument alpha=0.5 rend les points semi-transparents pour voir les zones de forte densité
sns.scatterplot(data=obs_gps, x='decimalLongitude', y='decimalLatitude', alpha=0.5, s=15, color="darkblue")
plt.title("Répartition spatiale des observations")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
