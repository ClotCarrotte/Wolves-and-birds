```python
import pandas as pd

# 1. Charger les données (le séparateur est une tabulation '\t')
df = pd.read_csv('occurrence.txt', sep='\t', low_memory=False)

# 2. Sélectionner uniquement les colonnes pertinentes pour alléger la mémoire
colonnes_utiles = [
    'scientificName', 'eventDate', 'decimalLatitude', 
    'decimalLongitude', 'individualCount', 'countryCode'
]
df_filtre = df[colonnes_utiles]

# 3. Exemple d'analyse : Compter le nombre d'observations par espèce
comptage_especes = df_filtre['scientificName'].value_counts()
print(comptage_especes.head())

# 4. Filtrer pour une espèce précise (ex: Corvus corax)
corbeaux = df_filtre[df_filtre['scientificName'].str.contains('Corvus corax', na=False)]
```

```python
# Chemin vers ton fichier
fichier = "occurrence.txt"

# Chargement des données dans un "DataFrame" (le format de tableau de pandas)
# sep="\t" est crucial : cela indique que le séparateur est une tabulation et non une virgule
# low_memory=False évite les problèmes de typage sur les fichiers contenant beaucoup de colonnes
tableau = pd.read_csv(fichier, sep="\t", low_memory=False)

# Afficher les 5 premières observations pour vérifier la structure
print(tableau.head())

# Afficher les dimensions totales (nombre de lignes et de colonnes)
print(f"\nLe tableau contient {tableau.shape[0]} observations et {tableau.shape[1]} colonnes.")
```
